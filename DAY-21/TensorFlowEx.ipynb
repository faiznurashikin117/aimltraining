# Find the shape,rank and size of the tensors

```python
#single number that means scalar
scalar = tf.constant(5)
scalar
```

```python
print("Scalar Information")
print("Shape:",scalar.shape)
print("Dimension:",scalar.ndim)
print("Rank:",tf.rank(scalar).numpy())
print("Size:",tf.size(scalar))
```

```python
#a number with direction
vector = tf.constant([10,10,20,30])
vector
```

```python
print("Vector Information")
print("Shape:",vector.shape)
print("Dimension:",vector.ndim)
print("Rank:",tf.rank(vector).numpy())
print("Size:",tf.size(vector))
```

```python
# a 2-D array of numbers
matrix = tf.constant([[10,20],[30,40]])
matrix
```

```python
print("Matrix Information")
print("Shape:",matrix.shape)
print("Dimension:",matrix.ndim)
print("Rank:",tf.rank(vector).numpy())
print("Size:",tf.size(vector))
```

#Converting Tensor into Numpy

```python
import numpy as np
numpy_Array = np.arange(1,25,dtype=np.int32)
numpy_Array
```

```python
tensor_numpy=tf.constant(numpy_Array,shape=(2,3,4))
tensor_numpy
```

```python
print("Tensor Information")
print("Shape:",tensor_numpy.shape)
print("Dimension:",tensor_numpy.ndim)
print("Rank:",tf.rank(tensor_numpy).numpy())
print("Size:",tf.size(tensor_numpy))
print("DataType of elements:",tensor_numpy.dtype)
print("Element along axis 0:",tensor_numpy.shape[0])
print("Element along axis 1:",tensor_numpy.shape[1])
print("Element along axis 2:",tensor_numpy.shape[2])
print("Total number of element: \t",tf.size(tensor_numpy))
```

```python
#Working with items of each dimension
tensor_numpy[:3,:3,:3]
```

```python
#Working with first elements of each dimension
tensor_numpy[:2,:2,:2]
```

**Arithmetic Operations with Tensors**

* Add
* Sub
* Multi
* Div


```python
tensor_one=tf.constant([[11,9],[2,5]])
tensor_one
```

```python
#tensor_one+15
tf.add(tensor_one,15)
```

```python
#tensor_one-5
tf.subtract(tensor_one,5)
```

```python
tensor_one*2
#tf.multiply(tensor_one,2)
```

```python
#tensor_one/2
tf.divide(tensor_one,2)
```

```python
tensor_new=tf.square(tensor_one)
tensor_new
```
