**Matrix Multiplication**

# We will use tf.matmul() method

**Rules:**
1.  Inner dimensions must match
    (m,n) @ (n,p)
    for e.g.
    * (5,7) @ (8,9) It will not work because n (7 and 8) in the dimension are different.
    * (5,8) @ (8,9) It will work because inner dimensions (7 and 8) dimension are same.

2. The result matrix has the shape of outer dimension means A (m,n) B(n,p), So, result has (m,p) e.g (4,7)@(7,5)->(4,5)

```python
import tensorflow as tf
import numpy as np
```

```python
tensor_A = tf.constant([[1,2],[3,4],[4,5]],dtype=tf.int32)
tensor_B = tf.constant([[5,6],[7,8]],dtype=tf.int32)
print(tensor_A)
print(tensor_B)
print("Shape of A:",tensor_A.shape)
print("Shape of B:",tensor_B.shape)
```

```python
tensor_C = tf.matmul(tensor_A,tensor_B)
print(tensor_C)
print("Shape of C:",tensor_C.shape)
```

```python
x = tf.constant([
                [1.,0.,-1],
                [2.,3.,1.]
                ],
                dtype=tf.float32)
print(x)
print("x shape:", x.shape)
```

```python
y = tf.constant([
    [4.],
    [5.],
    [6.]
    ],
    dtype=tf.float32)
y
```

```python
print("x shape:", x.shape)
print("y shape:", y.shape)
```

```python
z = tf.matmul(x,y)
z
```

```python
print(f"x-shape:{x.shape}, y-shape:{y.shape},z-shape:{z.shape}")
print(x.numpy())
print(y.numpy())
print(z.numpy())
```

**Transpose:**

Transpose swaps axes. for 2D matrix shape(m,n) becomes (n,m) after transpose

```python
tensor_one = tf.constant([
    [[1,2],[2,3]],
    [[4,5],[6,7]],
    ],dtype=tf.int32)
print(tensor_one)
print('Shape of tensor_one: \t',tensor_one.shape)
```

```python
tensor_one_transpose=tf.transpose(tensor_one)
print(tensor_one_transpose)
print("\nShape of tensor_one_transpose",tensor_one_transpose)
```

3D Transpose with explicit axis permutation

```python
tensor_two=tf.constant([
    [[1,2],[3,4]],
    [[5,6],[7,8]]
],dtype=tf.int32)
print(tensor_two)
```

```python
#swap first and last axes (0,1,2) ->(0,2,1)
t2_transpose=tf.transpose(tensor_two,perm=[0,2,1])
print(t2_transpose)
print("Shape of t2_transpose",t2_transpose.shape)
print("Shape of tensor_two",tensor_two.shape)
```

**Reshape**
# tf.reshape(tensor,new_shape)
# Change the shape without changing the data

**Rule**
* The total number of elements must stay the same


```python
tensor = tf.constant([
    [1,2,3],
    [4,5,6]
],dtype=tf.int32)
print(tensor)
print("Shape of tensor:",tensor.shape)
print("Number of elements are:",tf.size(tensor).numpy())
```

```python
tensor_r1=tf.reshape(tensor,shape=(3,2))
print(tensor_r1)
print("Shape of tensor:",tensor_r1.shape)
print("Number of elements are:",tf.size(tensor_r1).numpy())
```

```python
tensor_r2=tf.reshape(tensor,shape=(1,6))
print(tensor_r2)
print("Shape of tensor:",tensor_r2.shape)
print("Number of elements are:",tf.size(tensor_r2).numpy())
```

```python
tensor_r3=tf.reshape(tensor,shape=(6,1))
print(tensor_r3)
print("Shape of tensor:",tensor_r3.shape)
print("Number of elements are:",tf.size(tensor_r3).numpy())
```

```python
#infer dimension
tensor_r4=tf.reshape(tensor,shape=(-1,2))
print(tensor_r4)
print("Shape of tensor:",tensor_r4.shape)
print("Number of elements are:",tf.size(tensor_r4).numpy())
```
