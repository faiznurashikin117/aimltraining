**2. Clean the text**

```python
def clean_text(text):
  text = str(text)
  text = re.sub(r'http\S+','',text)
  text = text.translate(str.maketrans('','',string.punctuation))
  text = re.sub(r'\s+',' ',text).strip()
  text = re.sub(r'[^\x00-\x7f]','',text)
  text = text.lower()
  return text
```

```python
# data cleaning
df['clean_resume']=df['Resume'].apply(clean_text)
df.head()
```

```python
# 3,4,5
le=LabelEncoder()
df['Category']=le.fit_transform(df['Category'])
x = df['clean_resume']
y = df['Category']
df
```

```python
x_train,x_test,y_train,y_test=train_test_split(df['clean_resume'],df['Category'],test_size=0.2,random_state=42,stratify=y)
```

```python
tfidf = TfidfVectorizer (max_features = 5000,stop_words ='english',sublinear_tf=True)
x_train_tfidf = tfidf.fit_transform(x_train)
x_test_tfidf = tfidf.transform(x_test)
```

```python
model=LogisticRegression()
model.fit(x_train_tfidf,y_train)
```

```python
y_pred = model.predict(x_test_tfidf)
acc = accuracy_score(y_test,y_pred)
cnf_matrix = confusion_matrix(y_test,y_pred)
print("Accuracy:",acc)
print("\nConfusion matrix:\n",cnf_matrix)
classification_report = classification_report(y_test,y_pred)
print("\nClassification Report:\n",classification_report)
```

```python
new_resume=[
    "Organized and empathetic HR Manager with 3+ years of experience overseeing human resources duties including benefits management, onboarding and exit interviews. Increases employee satisfaction by advocating for expressed needs."
]
new_clean_resume = [clean_text(res) for res in new_resume]
new_clean_resume_vec = tfidf.transform(new_clean_resume)
predictions = model.predict(new_clean_resume_vec)
print("\n Predictions are:\n")
for res,prediction in zip(new_resume,predictions):
    print(f"{res} -----> {prediction}")
```
