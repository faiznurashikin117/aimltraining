# Supervised Learning

## Linear Regression

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('sales_data.csv')
data.head()
```

```python
x=data[['Advertising_Spend']]
y=data['Sales_Revenue']
```

```python
model = LinearRegression()
model.fit(x,y)
print('Training done')
```

```python
new_spend=float(input('Enter new spend to predict sales Revenue'))
predicted_revenue = model.predict([[new_spend]])
print(f'New Spend {new_spend} Predicted Revenue {predicted_revenue}')
```

```python
# import the required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
#1: collecting the data
df=pd.read_csv('sales_data.csv')
df.head(2)
```

```python
#2. splitting the data
adv_spend=df[['Advertising_Spend']]
sales_rev=df['Sales_Revenue']
```

```python
#3. model_creation
model = LinearRegression()
```

```python
#4. train the model
model.fit(adv_spend,sales_rev)
print('Trained successfully!')
```

```python
#5. predict the outcome
predicted_value=model.predict([[4000]])
print(f'New Spend {adv_spend} Predicted Revenue {predicted_value}')
```

# Exercise 1: Predict the data using Linear Regression - House Price Prediction
## Download house_price.csv file from github
## 

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('house_price.csv')  # reload the original file
df = df.replace('?', np.nan)
df = df.dropna(subset=['Price'])
df.head()
```

```python
x=df[['Size_sqft']]
y=df['Price']
```

```python
model=LinearRegression()
```

```python
model.fit(x,y)
```

```python
new_size=float(input('Enter your desired house size:'))
predicted_price=model.predict([[new_size]])
print(f'With house size {new_size}, the price is {predicted_price}')
               
```
