# LogisticReg

```python
# import the required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
```

```python
#1: collecting the data
df=pd.read_csv('pass_fail.csv')
df.head(2)
```

```python
#2. splitting the data
Hours_study=df[['Hours']]
Exam_result=df['Result']
```

```python
#3. model_creation
model = LogisticRegression()
```

```python
#4. train the model
model.fit(Hours_study,Exam_result)
print('Trained successfully!')
```

```python
pred_35=model.predict([[3.5]])
print('Predicted result for 3.5 hours:',pred_35[0])
```

```python
est_hour=float(input('Enter new hour:'))
pred_result=model.predict([[est_hour]])
print(f'If you study {est_hour}, then you will be {pred_result[0]}')
```

## Exercise 2


```python
# import the required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder
```

```python
#1: collecting the data
df=pd.read_csv('customer_purchase.csv')
df.head(2)
```

```python
#2. splitting the data
le=LabelEncoder()
df['Purchased'] = le.fit_transform(df['Purchased'])
x=df[['Age','Salary']]
y=df['Purchased']
```

```python
df
```

```python
#3. model_creation
model = LogisticRegression()
```

```python
#4. train the model
model.fit(x,y)
print('Trained successfully!')
```

```python
#5. Prediction

new_cust = [[28,35000]]
prediction = model.predict(new_cust)
print(f'Predicted Result(1=Yes, 0=No): {prediction [0]}')
```

```python
#5. Prediction

new_cust_age=int(input('Enter age:\t'))
new_cust_salary=int(input('Enter salary:\t'))
new_customer=[[new_cust_age,new_cust_salary]]
prediction=model.predict(new_customer)
if (prediction):
    print('Likely purchased')
else:
    print('Likely not purchased')
```

```python
# import the required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
```

```python
#1: collecting the data
df=pd.read_csv('spam_dataset.csv')
df
```

```python
#2. splitting the data
cv=CountVectorizer()
x = df['message_content']
x_cv=cv.fit_transform(x)
y=df['is_spam']
```

```python
#3. model_creation
model = LogisticRegression()
```

```python
#4. train the model
model.fit(x_cv,y)
print('Trained successfully!')
```

```python
#5. Prediction
sample = ["You have been selected to receive a special offer. Act fast!","Hi Alexandra,"]
prediction = model.predict(cv.transform(sample))
print(prediction)
```
