# KNN Algorithm in Supervised Machine Learning
## KNeighboursClassifier of sklearn library for this one

```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
# data collection and splitting the data
x = np.array([
    [1,3500],
    [2,3200],
    [3,3000],
    [4,2500],
    [5,2300],
    [6,2000],
    [7,2200],
    [2,3700]
])
y=['Unfit','Unfit','Unfit','Fit','Fit','Fit','Fit','Unfit']
```

```python
# create the model
model=KNeighborsClassifier(n_neighbors=3)
```

```python
# Train the model
model.fit(x,y)
```

```python
#predict
new_person=[[4.5,2000]]  #4.5 hrs exercise, 20000 calories
pred = model.predict(new_person)
print("Predicted Result is:\t", pred[0])
```

```python
#prediction from taking input from user
exercise_hours=float(input('Enter Number of Exercise Hours'))
cal_intake=float(input('Enter number of calories consumed'))
pred_fu=model.predict([[exercise_hours,cal_intake]])
print(f'If exercise done for {exercise_hours} hours, the calories intake for the day is {cal_intake}, you are likely {pred_fu}')
```

```python
import pandas as pd
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
```

```python
#1
data=pd.read_csv('student_grade.csv')
data
```

```python
#2
x=data[['Hours_Studied','Assignments_Completed']]
y=data['Grade']

#3
model=KNeighborsClassifier(n_neighbors=3)

#4
model.fit(x,y)

#5
hr_study=int(input('Enter duration of study in hour:\t'))
asg_compl=int(input('Enter amount of assignment completed:\t'))  
our_student=pd.DataFrame([[hr_study,asg_compl]],columns=['Hours_Studied','Assignments_Completed'])
pred_grade=model.predict(our_student)
print(f'If the student study for {hr_study} hours, and completed {asg_compl} assignments, they will likely be getting {pred_grade[0]}')
```
