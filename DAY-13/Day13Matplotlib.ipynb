```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.random.rand(10)
y=np.random.rand(10)
temp=np.random.randint(20,30,size=10)
plt.scatter(x,y,c=temp,cmap='plasma',s=120,edgecolor='black')
plt.colorbar(label='Temperature')
plt.title('Scatter plot colored by Temperature')
plt.xlabel('X-Random')
plt.ylabel('Y-Random')
plt.show()
```

```python
height_cm=[150,155,160,165,170,175,180,185]
weight_kg=[50,53,56,59,62,66,68,72]
age_years=[20,22,25,28,30,35,38,40]
```

```python
plt.figure(figsize=(8,4))
```

```python
scatter=plt.scatter(height_cm,weight_kg,c=age_years,cmap='rainbow',s=120,alpha=0.9)
plt.colorbar(scatter,label='Age(Years)')
plt.title('Scatter plot: Height vs Weight (colored by Age)')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.grid(color='blue',linestyle='--',linewidth='0.7')
plt.show()
```

# Pie Plot Example
## plt.pie(x,labels='',colors='',autopct='',startangle='',shadow=,explode=radius)

```python
depts=['HR','Web-Development','App-Development','Agile-Scrum','Testing']
expenses=[12000,25000,18000,15000,20000]
```

```python
plt.figure(figsize=(7,7))
```

```python
plt.pie(expenses,labels=depts,autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0,0),shadow=True,radius=1.0)
plt.title('Department Wise Expenses Report')
plt.show()
```

```python
plt.pie(expenses,labels=depts,colors=plt.cm.rainbow([0.1,0.3,0.5,0.7,0.9]),autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0,0),shadow=True,radius=1.0)
plt.title('Department Wise Expenses Report')
plt.show()
```

```python
plt.pie(expenses,labels=depts,colors=plt.cm.magma([0.1,0.3,0.5,0.7,0.9]),autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0,0),shadow=True,radius=1.0)
plt.title('Department Wise Expenses Report')
plt.show()
```

```python
plt.pie(expenses,labels=depts,colors=plt.cm.inferno([0.1,0.3,0.5,0.7,0.9]),autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0,0),shadow=True,radius=1.0)
plt.title('Department Wise Expenses Report')
plt.show()
```

# create a pie plot using following Data
## sales_data={'Mobile':45000,'Clothing'=30000,'Groceries':35000,'Home Decor':2000}
## show percentage with two decimal
## use colormap to give the different color to your pie

```python
sales_data={'Mobile':45000,'Clothing':30000,'Groceries':35000,'Home Decor':20000}
categories=list(sales_data.keys())
product_sales=list(sales_data.values())
plt.figure(figsize=(7,7))
plt.pie(product_sales,labels=categories,colors=plt.cm.magma([0.1,0.3,0.5,0.7,0.9]),textprops={'color':'yellow','fontsize':10,'fontweight':'bold'},autopct='%1.2f%%',startangle=140,explode=(0,0.05,0,0),shadow=True,radius=1.0)
plt.title('Product Sales Report')
#plt.savefig('filenamewtihextension',dpi=value,bbox_inches='')
plt.show()
```

# Histogram Example

```python
marks=[10,20,30,35,45,55,78,90,34,12,80,90,45,100,0,50,50,12,56,90,80,11,7]
plt.figure(figsize=(8,5))
plt.hist(marks,bins=5,color='skyblue', edgecolor='darkblue',alpha=0.8)
plt.title('Histogram for student marks')
plt.xlabel('Marks range')
plt.ylabel('Number of students')
plt.grid(color='blue',linestyle='--',linewidth='0.7')
plt.show()
```
