```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'April']
sales = [2000, 3000, 2500, 3600]
```

# plt.plot(x, y, color='color name', linestyle='line_style', linewidth=value, marker='marker symbol', label='label name')

```python
plt.plot(months,sales,color='blue',linestyle='-.',linewidth='2',marker='x',label='Monthly Sales Data')
plt.title("Monthly Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales per Months")
plt.grid(True)
plt.legend(fontsize='15')
plt.show()
```

```python
plt.plot(months,sales,color='blue',linestyle='-.',linewidth='2',marker='p',label='Monthly Sales Data')
plt.title("Monthly Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales per Months")
plt.xlim(0,4)
plt.ylim(2000,4000)
plt.grid(True)
plt.legend(fontsize='15')
plt.show()
```

```python
import numpy as np
```

```python
x=np.linspace(0,2,100)
plt.plot(x,x,label='Linear')
plt.plot(x,x**2,label='Quadratic')
plt.plot(x,x**3,label='Cubic')
plt.xlabel('X Lines')
plt.ylabel('Y Lines')
plt.title('Simple Line Plot')
plt.legend(fontsize='15')
plt.show()
```

# Bar Plot Example
## plt.bar(x,y,color='',edgecolor='')

```python
categories=['Electronics','Mobiles','Book Categories','Groceries','Home Decorations']
sales=[5000,6000,3000,9000,4000]
plt.figure(figsize=(8,4))
color_bar=('red', 'yellow','magenta','skyblue','green')
plt.bar(categories,sales,color=color_bar,edgecolor='blue')
plt.xlabel('categories')
plt.ylabel('Sales Values (MYR)')
plt.title('Categories wise Sales Report')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
categories=['Electronics','Mobiles','Book Categories','Groceries','Home Decorations']
sales=[50000,60000,30000,90000,40000]
plt.figure(figsize=(8,4))
plt.bar(categories,sales,color='skyblue',edgecolor='blue',label='Sales Info')
plt.xlabel('categories')
plt.ylabel('Sales Values (MYR)')
plt.title('Categories wise Sales Report')
plt.xticks(rotation=30,fontsize=12,color='magenta')
plt.yticks([20000,40000,60000,80000,100000],['20K','40K','60K','80K','100K'],fontsize=12,color='green')
plt.legend()
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

# Scatter Plot

## plt.scatter(x,y,s='',c='colorName',marker='markername',alpha=transparency value,edgecolor='color',label='label of legend'

```python
height_cm=[150,155,160,165,170,175,180,185]
weight_kg=[50,53,56,59,62,66,68,72]
plt.figure(figsize=(8,5))
plt.scatter(height_cm,weight_kg,s=100,color='pink',marker='v',alpha=0.9,edgecolor='black',label='weight height compr')
plt.title('Scatter Plot Height vs Weight')
plt.xlabel('Height(cm)')
plt.ylabel('Weight(kg)')
plt.xlim(140,190)
plt.ylim(40,90)
plt.xticks(rotation=30,fontsize=12,color='blue')
plt.yticks([50,55,60,65,70,75,80,85],['50kg','55kg','60kg','65kg','70kg','75kg','80kg','85kg'],fontsize=12,color='green')
plt.grid(color='lightblue',alpha=0.5,linestyle='-.',linewidth=0.5)
plt.legend()
plt.show()
```
