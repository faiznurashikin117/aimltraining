```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("netflix_data.csv")
print(df.head())
```

```python
df=df.dropna(subset=['release_year'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
plt.figure(figsize=(10,5))
plt.hist(df['release_year'],bins=100,color='green', edgecolor='darkblue',alpha=0.8)
plt.title('Netflix Release Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue',linestyle='--',linewidth='0.7')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df=pd.read_csv("netflix_data.csv")
df=df.dropna(subset=['release_year'])
df['release_year']=df['release_year'].astype(int)
plt.figure(figsize=(10,5))
plt.hist(df['release_year'],bins=100,color='green', edgecolor='pink',alpha=0.8)
plt.title('Netflix Release Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(color='blue',linestyle='--',linewidth='0.7')
plt.tight_layout()
plt.savefig('Netflix Release Titles by Year.pdf',dpi=300,bbox_inches='tight')
plt.show()
```

```python
df=df.dropna(subset=['release_year'])
```

```python
df['release_year']=df['release_year'].astype(int)
```

```python
movies=df[df['type']=='Movie']
shows=df[df['type']=='TV Show']
print(movies,shows)
```

```python
plt.figure(figsize=(10,5))
plt.hist([movies['release_year'],shows['release_year']],bins=30,color=['green','red'], edgecolor='pink',alpha=0.8,label=['Movies','TV Shows'])
plt.title('Netflix Release Titles by Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.grid(axis='y',color='blue',linestyle='--',linewidth='0.7')
plt.legend(title='Movies/TV Shows')
plt.tight_layout()
plt.savefig('Netflix shows and movies.jpg',dpi=300,bbox_inches='tight')
plt.show()
```

# Use netflix_data.csv dataset
## print how many movies and TV Shows are available on Netflix
### Display movies and TV Shows proportion in a pie plot
#### Save file as pdf

```python
df=df.dropna(subset=['release_year'])
df['release_year']=df['release_year'].astype(int)
movies=df[df['type']=='Movie']
shows=df[df['type']=='TV Show']
num_movies = len(movies)
num_shows = len(shows)
values=[num_movies,num_shows]
categories=['Movie','TV Show']
plt.figure(figsize=(7,7))
plt.pie(values,labels=categories,colors=plt.cm.magma([0.1,0.3]),textprops={'color':'yellow','fontsize':10,'fontweight':'bold'},autopct='%1.2f%%',startangle=140,explode=(0,0.05),shadow=True,radius=1.0)
plt.title('Netflix Shows Count')
#plt.savefig('filenamewtihextension',dpi=value,bbox_inches='')
plt.savefig('Netflix Shows Count.pdf',dpi=300,bbox_inches='tight')
plt.show()
```
