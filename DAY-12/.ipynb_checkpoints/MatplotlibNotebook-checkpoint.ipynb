# Basics of matplotlib chart
    ## pyplot: interface for creating and managing plot
    ## Figure: the complete window to create plot
    ## Axes: the area where data is plotted
    ## Axis: x-line and y-line to define scale and ticks
    ## Plot: the graph of data

```python
plt.figure(figsize=(8,6))
```

```python
plt.plot(x,y,marker='D',linestyle='-')
```

```python
plt.plot(x,y,marker='x',linestyle='-')
```

```python
plt.plot(x,y,marker='o',linestyle='-')
```

# First example to create a line chart from our data
    ## students.xlsx for fetching the data
    ## Generate line chart based on Student name and Student Marks

```python
import pandas as pd
```

```python
df=pd.read_excel("C:/Users/Faridah/Desktop/AIML/Training Logbook/DAY-11/students.xlsx")
print(df)
```

```python
names=df['Name']
marks=df['Marks']
```

```python
plt.figure(figsize=(4,3))
plt.plot(names,marks,marker='s',color='green')
plt.title('Student Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Student Marks')
plt.grid(True)
plt.show()
```

```python
print('Changed x and y axis')
plt.figure(figsize=(5,3))
plt.plot(marks,names,marker='o',color='orange')
plt.title('Student Progress Chart')
plt.xlabel('Student Marks')
plt.ylabel('Student Name')
plt.grid(False)
plt.show()
```

```python
print('Solid Line')
plt.figure(figsize=(4,3))
plt.plot(names,marks,marker='s',color='green',linestyle='-')
plt.title('Student Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Student Marks')
plt.grid(True)
plt.show()
```

```python
print('Dashed Line')
plt.figure(figsize=(4,3))
plt.plot(names,marks,marker='o',color='blue',linestyle='--')
plt.title('Student Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Student Marks')
plt.grid(True)
plt.show()
```

```python
print('Dotted Line')
plt.figure(figsize=(4,3))
plt.plot(names,marks,marker='x',color='green',linestyle=':')
plt.title('Student Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Student Marks')
plt.grid(False)
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.plot(names,marks,marker='o',color='blue',linestyle='-.')
plt.title('Student Marks Line Chart')
plt.xlabel('Student Name')
plt.ylabel('Student Marks')
plt.grid(True)
plt.show()
```
