# Scatter Plots

## In scatter plots we represent data in points, where each point represent a data observation, We can check how to values correlate

```python
product_sales={'Laptop':750,'Smart Phone':650,'Smart Watch':400,'Tablets':500,'AC':400}
products=list(product_sales.keys())
sales=list(product_sales.values())
plt.scatter(products,sales)
plt.title('Basic Product and Sales Representation')
plt.xlabel('Products')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```

```python
players = ['MSD', 'Rohit', 'Virat', 'Gill', 'Shami']
match1_score = [45, 67, 56, 78, 89]
match2_score = [55, 78, 40, 80, 97]
match3_score = [45, 74, 44, 88, 94]
plt.figure(figsize=(8,5))
plt.scatter(players, match1_score, color='red',label='Match One',s=100,alpha=0.3)
plt.scatter(players, match2_score, color='blue',label='Match Two',s=100,marker='x',alpha=0.3)
plt.scatter(players, match3_score, color='green',label='Match Three',s=100,marker='D',alpha=0.3)
plt.title('Comparison of players in two matches')
plt.xlabel('Players')
plt.ylabel('Scores')
plt.legend(loc='best')
plt.grid(True,linestyle='--',alpha=0.5)
plt.show()
```

# Create a bar chart using following data
## we have the following libraries 
    ## python_lib=['Numpy','Pandas','Matplotlib','Scikit-learn']
## Liked by users as follows
    ## liked_by=[5000,7000,9000,4000]
## Create a vertical bar graph where each bar represents the value of python_lib respective liked by users of the libraries
    ##every bar must have different colors

```python
python_lib=['Numpy','Pandas','Matplotlib','Scikit-learn']
liked_by=[5000,7000,9000,4000]
plt.figure(figsize=(6,3))
bar_colors=['purple','red','green','yellow',]
plt.bar(python_lib,liked_by,color=bar_colors,width=0.5,edgecolor='black')
plt.title('Popularity of Python Libraries Among Users')
plt.xlabel('Python Libraries')
plt.ylabel('Popularity Score')
plt.show()
```

```python
python_lib=['Numpy','Pandas','Matplotlib','Scikit-learn']
liked_by=[5000,7000,9000,4000]
plt.figure(figsize=(6,3))
bar_colors=['purple','red','green','yellow',]
plt.barh(python_lib,liked_by,color=bar_colors,edgecolor='black')
plt.title('Popularity of Python Libraries Among Users')
plt.xlabel('Python Libraries')
plt.ylabel('Popularity Score')
plt.show()
```
