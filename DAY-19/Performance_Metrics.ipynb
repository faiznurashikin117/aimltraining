# MAE : Mean Absolute Error

```python
from sklearn.metrics import mean_absolute_error
```

```python
actual_values=[70,80,90,85,95]
predicted_values=[68,82,88,80,97]
```

```python
mae=mean_absolute_error(actual_values,predicted_values)
print('MAE', mae)
```

# MSE : Mean Squared Error

```python
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
house_size = np.array ([[1000],[1500],[2000],[2500],[3000]])
house_price_thousand = np.array([120,180,250,310,370])
```

```python
model = LinearRegression()
model.fit(house_size,house_price_thousand)
```

```python
y_pred = model.predict(house_size)
```

```python
mse = mean_squared_error(house_size,y_pred)
print('MSE for this model is: ',mse)
print(mse**2)
```

```python
rmse = root_mean_squared_error(house_size,y_pred)
print('Root Mean squared Error:',rmse)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(house_size, house_price_thousand, color='blue', label='Actual Values', s=100)
plt.plot(house_size,y_pred, color='red', label='Predicted Values', linewidth=2)
# Draw vertical lines (residuals)
for i in range(len(house_size)):
    plt.vlines(x=house_size[i], ymin=min(house_price_thousand[i], y_pred[i]),
               ymax=max(house_price_thousand[i], y_pred[i]), color='green', linestyle='dashed')

plt.xlabel('House Size (sqft)')
plt.ylabel('Price (in thousand $)')
plt.title('Actual vs Predicted House Prices')
plt.legend()
plt.show()
```

# RMSE : Root Mean Squared Error
