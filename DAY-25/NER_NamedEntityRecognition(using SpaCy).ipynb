```python
import spacy
nlp=spacy.load("en_core_web_sm")
print('SpaCy Loadded Succcessfully')
```

```python
text = "Bill Gates founded Microsoft in Year 1994 in California -USA"
```

```python
doc = nlp(text)
for ent in doc.ents:
  print(ent.text,ent.label_,spacy.explain(ent.label_))
```

```python
from spacy import displacy
displacy.render(doc,style='ent',jupyter=True)
```

```python
corpus = [
    "TajMahal is located in Agra India",
    "Google was founded by Larry Page",
    "Twin Towers is located in Kuaala Lumpur",
    "Discovery of India was written by Pt. Nehru"
]
```

```python
for text in corpus:
  doc = nlp(text)
  for ent in doc.ents:
    print(ent.text,ent.label,spacy.explain(ent.label_))
```
