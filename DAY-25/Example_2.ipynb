```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("/content/UpdatedResumeDataSet.csv",encoding='utf-8')
df.shape
df.head()
```

```python
print("\n Columns",df.columns.tolist())
```

```python
sample_df = df.sample(10,random_state=42).reset_index(drop=True)
sample_df
```

```python
sample_df=sample_df.rename(columns={'Resume':'Resume_Text'})
sample_df
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix=vectorizer.fit_transform(sample_df['Resume_Text'])
tfidf_matrix.shape
```

```python
similaritiness_matrix = cosine_similarity(tfidf_matrix)
sim_df = pd.DataFrame(similaritiness_matrix, index=sample_df['Category']+"_"+sample_df.index.astype(str),
                      columns=sample_df['Category']+"_"+sample_df.index.astype(str))
sim_df.head()
sim_df
```

```python
resumes = {
    "candidates":['Sam','Arsh','Maan'],
    "resume_text":["Python developer skilled in Django, Flask, Rest API and MySQL","Machine learning engineer experience in Python,TensorFlow, NLP and Data Analysis","Full Stack developer ith .net,javascript, React, Node.js and MongoDB experience"]
}
```

```python
df = pd.DataFrame(resumes)
df
```

```python
vectorize = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorize.fit_transform(df['resume_text'])
tfidf_matrix.shape
tfidf_matrix
```

```python
similaritiness_matrix = cosine_similarity(tfidf_matrix)
sim_df = pd.DataFrame(similaritiness_matrix, index=df["candidates"],columns=df["candidates"])
sim_df.round(2)
```
