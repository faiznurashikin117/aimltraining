**NLP Pipeline**

```python
# import Required Libraries
import nltk
import spacy  #nlp libraries
from nltk.corpus import stopwords   #analyze data
from nltk.tokenize import word_tokenize    #
from nltk.stem import PorterStemmer, WordNetLemmatizer
import re
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')   #to handle multilingual words
nltk.download('punkt_tab')
```

```python
our_text = "Artificial Intelligence and Machine Learning are going to change the future of technology!!"
```

```python
# splits text into individual words/tokens
tokens = word_tokenize(our_text)
tokens
```

```python
# cleaning
# Lowercasing makes them consistent and easier to process
# Lowercasing helps the model focus on meaning, not casing
# reduce confusion/duplicate word treated as different word
clean_tokens = [re.sub(r'[^a-zA-Z]','',w.lower()) for w in tokens if w.isalpha()]
clean_tokens
```

```python
# Stop Word Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [token for token in clean_tokens if token not in stop_words]
# remove common less meaningful words
filtered_tokens
```

```python
# Stemming
stemmer = PorterStemmer()
stemmed_tokens = [stemmer.stem(token) for token in filtered_tokens]
# Reduce the words to their root form( not always a word from dictionary)
stemmed_tokens
```

```python
lemmetizer = WordNetLemmatizer()
lemmatized_tokens = [lemmetizer.lemmatize(token) for token in filtered_tokens]
# Convert tokens to their dictionary form (lemma) using POS: part of speech (meaningful words recovery)
lemmatized_tokens
```

```python
from nltk import corpus
corpus = [
    "Artificial Intelligence is transforming industries",
    "Machine Learning is a brunch of artificial intelligence",
    "Deep Learning is a subfield of machine Learning",
    "Natural Language Processing is a subfield of artificial intelligence",
]
```

```python
cv = CountVectorizer()
bow_matrix = cv.fit_transform(corpus)
print("********Features Name(BOW)********")
print(cv.get_feature_names_out())
print("\n********BOW Matrix Array********")
print(bow_matrix.toarray())
```
