# Matplotlib.pyplot.subplots()
## subplot(): provides a way to plot multiple plots on single figure
## Given number of rows and columns. It returns a tuple (fig,ax) giving a single figure fig with an array of axes of ax.

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
```

```python
sales_data={'Electronics':45000,'Clothing':30000,'Grocery':25000,'Mobile':20000}
category=list(sales_data.keys())
actual_sale=list(sales_data.values())
fig,((ax1,ax2),(ax3,ax4))=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
x=np.random.random(10)
ax1.plot(x,x/2)
ax2.scatter(np.random.random(10),np.random.random(10))
ax3.bar(category,actual_sale)
ax4.hist(np.random.randn(1000))
```

```python
our_lines=np.linspace(0,10,100)
```

```python
product_prices={'Hair Oil':20,'Face Wash':30,'Lip Balm':10,'Face Cream':15}
```

```python
our_numbers=np.random.random(10)
```

```python
our_dots=np.random.randn(100)
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
ax[0,0].plot(our_lines,our_lines/2)
ax[0,1].bar(list(product_prices.keys()),list(product_prices.values()),color=['blue','green','red','orange'])
ax[1,0].scatter(np.random.randn(100),np.random.randn(100))
ax[1,1].hist(our_dots,color='red')
plt.tight_layout()
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,5))
ax[0,0].plot(our_lines,our_lines/2)
ax[0,1].bar(list(product_prices.keys()),list(product_prices.values()),color=['blue','green','red','orange'])
ax[1,0].scatter(np.random.randn(100),np.random.randn(100),color='pink')
ax[1,1].hist(our_dots,color='red')

# first row, 0 & first column, 0 Line Plot
ax[0,0].set_title('Line Plot Example')
ax[0,0].set_xlabel('X axis')
ax[0,0].set_ylabel('Y axis')
ax[0,0].grid(color='blue',linestyle='--',linewidth='0.7')

# first row, 0 & second column, 1 Bar Chart
ax[0,1].set_title('Product Sales Information')
ax[0,1].set_xlabel('')
ax[0,1].set_ylabel('')
ax[0,1].grid(color='blue',linestyle='--',linewidth='0.7')

# second row, 1 & first column, 0 Scatter Plot
ax[1,0].set_title('Our Scatter Plot')
ax[1,0].set_xlabel('Product Name')
ax[1,0].set_ylabel('Product Sale')
ax[1,0].grid(color='blue',linestyle='--',linewidth='0.7')

# second row, 1 & second column, 1 Histogram
ax[1,1].set_title('Histogram Example')
ax[1,1].set_xlabel('Value Range')
ax[1,1].set_ylabel('Frequency')
ax[1,1].grid(color='blue',linestyle='--',linewidth='0.7')

#.....................................
fig.suptitle('Matplot Lib SubPlots Example using 2 rows and 2 columns')
plt.tight_layout()
plt.savefig('SubplotEx',dpi=300,bbox_inches='tight')
```
