```python
import pandas as pd
```

```python
data= { 'Students':['Amit','Anam','Chi','Sam','Ravi','Jai'],'Marks':[85,None,65,75,None,55]}
df=pd.DataFrame(data)
df
```

```python
print("Missing_Data")
print("Number of Missing values are:",df.isnull().sum())
```

```python
average_marks=df["Marks"].mean()
print(average_marks)
```

```python
df.fillna(average_marks, inplace=True)
```

```python
print("after fillna() call")
df
```

```python
## Solution1: find out number of missing data
df=pd.read_csv("students_performance_dirty.csv")
df.head()
print('Number of Missing Values are:',df.isnull().sum())
rows, cols = df.shape
print(f"Number of rows: {rows}")
print(f"Number of columns: {cols}")

## Solution2: drop rows where gender is missing
df.dropna(subset=['gender'],inplace=True)
#df is updated with no missing gender values

## Solution3: fill study_hours with average_study_hours
df.fillna({'study_hours':df['study_hours'].mean()},inplace=True)
#df is replaced with average study hours

rows, cols = df.shape
print("After cleaning:")
print(f"Number of rows: {rows}")
print(f"Number of columns: {cols}")

df
```
