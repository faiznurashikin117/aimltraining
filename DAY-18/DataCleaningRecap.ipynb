```python
import pandas as pd
```

```python
# Sample dataset
data = {
    'Name': ['Ali','Sara','John', None, 'Mary'],
    'Age': [25, None, 30, 22, None],
    'City': ['KL','penang', None,'Johor','KL']
}
```

```python
df = pd.DataFrame(data)
print('Original Dataframe:')
print(df)
```

## check the missing values Row by Row

```python
for index, row in df.iterrows():
    print(f'Row {index}:Missing = {row.isnull().any()}, Details = {row.isnull().to_dict()}')
```

```python
# show only rows with missing data
missing_rows = df[df.isnull().any(axis=1)]
print('Rows with missing data')
print(missing_rows)
```

```python
# Stimulate rows that would be dropped using dropna()
to_drop = df[df.isnull().any(axis=1)]
print('These rows will be dropped using dropna():')
print(to_drop)
```

```python
df_cleaned = df.dropna()
print('After dropna():')
print(df_cleaned)
```

```python
print('Before filling:')
print(df)
```

```python
df_filled = df.fillna({
    'Name' : 'Unknown',
    'Age' : df['Age'].mean(),
    'City' : 'Not Available'
})
```

```python
print('After filling:')
print(df_filled)
```

```python
df_sfill = df.fillna({
    'Name' : 'Unknown',
    'Age' : round(df['Age'].mean(),2),
    'City' : 'Not Available'
})
print('If want to make data in 2 decimal places only')
print(df_sfill)
```

```python
import math
```

```python
df_sfill = df.fillna({
    'Name' : 'Unknown',
    'Age' : math.floor(df['Age'].mean()),
    'City' : 'Not Available'
})
print('If want to make data without decimal places')
print(df_sfill)
```

```python
df_sfill = df.fillna({
    'Name' : 'Unknown',
    'Age' : int(df['Age'].mean()),
    'City' : 'Not Available'
})
print('If want to make data without decimal places')
print(df_sfill)
```
