```python
print("All greater than 99?",np.all(sorted_arr>99))
```

# Slicing 1 D Array
    1. To take elements from starting index to end index
    2. e.g.[start:end]

```python
print(our_arr)
```

```python
sorted_arr=np.sort(our_arr)
print(np.sort(our_arr))
```

## starting from an index 1  to 2

```python
slice_sorted_arr=sorted_arr[1:3]
print(slice_sorted_arr)
```

## starting from an index 3 going to the end

```python
print(sorted_arr[3:])
```

## starting from beginning index going to the forth(0,1,2,3)

```python
print(sorted_arr[:4])
```

```python
print(sorted_arr[:5])
```

## step
    ## [start:end step]
    ## step: every step element

```python
our_slice=sorted_arr[1:6:2]
print(our_slice)
```

```python
print("Original Array")
print(sorted_arr)
print("Every Third Element")
every_third_element=sorted_arr[::3]
print(every_third_element)
```

# Slice Reverse

```python
print("Original Array")
print(sorted_arr)
print("Reverse Array:")
print(sorted_arr[::-1])
```

```python
print("Last Five Element")
print(sorted_arr[-5:])
```

```python
print("Slicing Done")
```
